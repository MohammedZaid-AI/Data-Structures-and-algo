```python
def linear_search(arr,item):

  for i in range(len(arr)):
    if arr[i]==item:
      return i

  return -1
```

```python
arr=[24,34,76,87,12,44,22]

linear_search(arr,44)
```

```python
def binary_search(arr,low,high,item):

  if low<=high:
    mid=(low+high)//2

    if arr[mid]==item:
      return mid
    elif arr[mid]>item:
      return binary_search(arr,low,mid-1,item)
    else:
      return binary_search(arr,mid+1,high,item)
  else:
    return -1
```

```python
def is_sorted(arr):

  sorted=True

  for i in range(len(arr)-1):
    if arr[i]> arr[i+1]:
      sorted = False

  return sorted
```

```python
import random
def monkey_sort(arr):

  while not is_sorted(arr):
    random.shuffle(arr)
    print(arr)

  print(arr)
```

```python
monkey_sort([2,1,5,35,12,10])
```

```python
def bubble_sort(arr):

  for i in range(len(arr)-1):
    flag=0
    for j in range(len(arr)-1-i):
      if arr[j]>arr[j+1]:
        arr[j],arr[j+1]=arr[j+1],arr[j]
        flag+=1
    if flag==0:
      break

  print(arr)
```

```python
bubble_sort([7,4,88,75,66,22,53,25])
```
