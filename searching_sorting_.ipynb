```python
def linear_search(arr,item):
  for i in range(len(arr)):
    if arr[i]==item:
      return i

  return -1
```

```python
arr=[3,6,2,9,5,8]
```

```python
linear_search(arr,5)
```

```python
def binary_search(arr, low, high, item):

  if low<=high:
    mid=(high+low)//2

    if arr[mid]==item:
      return mid
    elif arr[mid]>item:
      return binary_search(arr,low,mid-1,item)
    else:
      return binary_search(arr,mid+1,high,item)

  else:
    return -1
```

```python
def is_sorted(arr):
  sorted=True

  for i in range(len(arr)-1):
    if arr[i]>arr[i+1]:
      sorted=False

  return sorted
```

```python
is_sorted(arr)
```

```python
import time
import threading

def sleep_and_print(num):
    time.sleep(num)
    print(num)

def sleep_sort(arr):
    threads = []

    for num in arr:
        thread = threading.Thread(
            target=sleep_and_print,
            args=(num,)
        )

        thread.start()
        threads.append(thread)

    for thread in threads:
        thread.join()


arr = [5, 2, 4, 1, 3]
sleep_sort(arr)
```

```python
def bubble_sort(arr2):
  for i in range(len(arr2)-1):
    for j in range(len(arr2)-1-i):
      if arr2[j]>arr2[j+1]:
        temp=arr2[j]
        arr2[j]=arr2[j+1]
        arr2[j+1]=temp

  print(arr2)
```

```python
arr2=[3,5,7,1]
```

```python
bubble_sort(arr2)
```
