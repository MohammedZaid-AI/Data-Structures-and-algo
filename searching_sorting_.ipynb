```python
def linear_search(arr,item):
  for i in range(len(arr)):
    if arr[i]==item:
      return i

  return -1
```

```python
arr=[3,6,2,9,5,8]
```

```python
linear_search(arr,5)
```

```python
def binary_search(arr, low, high, item):

  if low<=high:
    mid=(high+low)//2

    if arr[mid]==item:
      return mid
    elif arr[mid]>item:
      return binary_search(arr,low,mid-1,item)
    else:
      return binary_search(arr,mid+1,high,item)

  else:
    return -1
```
