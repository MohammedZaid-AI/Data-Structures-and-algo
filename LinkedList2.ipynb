```python
class Node:

  def __init__(self,value):
    self.data=value
    self.next=None

class LinkedList:

  def __init__(self):
    self.head=None
    self.n=0

  def __len__(self):
    return self.n

  def insert_head(self,value):
    new_node=Node(value)
    new_node.next=self.head
    self.head=new_node
    self.n+=1

  def __str__(self):
    curr=self.head
    result=''
    while curr!=None:
      result=result + str(curr.data) +'->'
      curr=curr.next
    return result[0:-2]

  def append(self, item):
      new_node = Node(item)

      if self.head ==None:
        self.head=new_node
        self.n+=1
        return

      curr = self.head
      while curr.next != None:
          curr = curr.next
      curr.next = new_node

      self.n += 1

  def insert_after(self,after,value):
    new_node=Node(value)

    curr=self.head

    while curr!=None:
      if curr.data==after:
        break
      curr=curr.next

    if curr!=None:
      new_node.next=curr.next
      curr.next=new_node
    else:
        return 'Item not found'
    self.n+=1


  def clear(self):
    self.head=None
    self.n=0

  def delete_head(self):
    if self.head !=None:
      self.head=self.head.next
      self.n-=1
    else:
      return "Empty"

  def delete_tail(self):
    if self.head==None:
      return "empty"
    if self.head.next==None:
      self.delete_head()
      return

    curr=self.head
    while curr.next.next!=None:
      curr=curr.next
    curr.next=None
    self.n-=1

  def delete_value(self,value):
    if self.head.data==value:
      self.delete_head()
      return
    curr=self.head
    while curr!=None:
      if curr.next.data==value:
        break
      curr=curr.next

    if curr!=None:
        curr.next=curr.next.next
        self.n-=1
        return
    return "Item not found"


  def search(self,value):
    curr=self.head
    pos=0

    while curr!=None:
      if curr.data==value:
        break
      curr=curr.next
      pos+=1

    if curr==None:
      return "Item not found"
    return pos

    if curr!=None:
      return pos
    return -1

  def __getitem__(self,index):
    curr=self.head
    pos=0

    while curr!=None:
      if pos==index:
        break
      curr=curr.next
      pos+=1

    if curr==None:
      return "Item not found"
    return curr.data


  def replace_max(self,value):
    temp=self.head
    max=temp

    while temp!=None:
      if temp.data > max.data:
        max=temp
      temp=temp.next

    max.data=value

  def sum_odd(self):
    temp=self.head
    counter=0
    sum=0

    while temp!=None:
      if counter %2!=0:
        sum=sum+temp.data
      counter+=1
      temp=temp.next

    print(sum)

  def reverse(self):
    prev_node=None
    current_node=self.head

    while current_node!=None:
      new_node=current_node.next
      current_node.next=prev_node
      prev_node=current_node
      current_node=new_node

    self.head=prev_node

  def change_sent(self):
    temp=self.head

    while temp!=None:
      if temp.data=="*" or temp.data=="/":
        temp.data=" "

        if temp.next.data=="*" or temp.next.data=="/":
          temp.next.next.data=temp.next.next.data.upper()
          temp.next=temp.next.next
      temp=temp.next
```
