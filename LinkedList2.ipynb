```python
class Node:

  def __init__(self,value):
    self.data=value
    self.next=None

class LinkedList:

  def __init__(self):
    self.head=None
    self.n=0

  def __len__(self):
    return self.n

  def insert_head(self,value):
    new_node=Node(value)
    new_node.next=self.head
    self.head=new_node
    self.n+=1
```

```python
LL=LinkedList()
n1=LL.insert_head(10)
LL.insert_head(20)
LL.insert_head(30)
LL.insert_head(40)
LL.insert_head(50)
```
