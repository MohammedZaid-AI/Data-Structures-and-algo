Think of this code as a smart search. Imagine you are at a party, and you need to find two people whose ages add up to exactly 50.
Instead of asking every person to pair up with every other person (which would take forever), you walk through the room once with a notebook.
------------------------------
## The "Notebook" Strategy (Line-by-Line)

   1. num_to_index = {} (The Notebook)
   * This is your empty notebook (a Python dictionary).
      * You’ll use it to write down: "I saw the number X at position Y."
   2. for i, num in enumerate(arr): (Walking the Room)
   * You walk through the list one by one.
      * num is the number you are looking at right now.
      * i is its "seat number" (index) in the list.
   3. complement = target - num (The Missing Piece)
   * If your goal is 10 and the current number is 7, you know you need a 3 to win.
      * complement is that "missing piece" ($10 - 7 = 3$).
   4. if complement in num_to_index: (Checking the Notebook)
   * This is the most important part. You check your notebook: "Have I already seen a 3 earlier in this walk?"
      * Because it's a HashMap/Dictionary, Python can check this almost instantly, no matter how big the notebook is.
   5. return [num_to_index[complement], i] (The Success)
   * If you did see a 3 earlier, you’re done!
      * You return the seat number of the 3 (which you stored in the notebook) and your current seat number i.
   6. num_to_index[num] = i (Writing it Down)
   * If you haven't seen the "missing piece" yet, you don't give up.
      * You write your own number and seat number in the notebook: num_to_index[7] = index. Now, if someone later needs a 7, they will find you in the notebook!
   
------------------------------
## Why is this better than a normal search?
If you didn't have the notebook, you would have to:

   1. Pick the first number.
   2. Scan the entire rest of the list for its partner.
   3. Pick the second number.
   4. Scan the entire rest of the list again.

That's like looking at every person in the room over and over. With the Dictionary (HashMap), you only look at each person once.
## A Quick Example
List: [2, 7, 11] | Target: 9

* Step 1 (Number 2): Target is 9. I need a 7. Is 7 in my notebook? No.
* Notebook now says: {2: 0}
* Step 2 (Number 7): Target is 9. I need a 2. Is 2 in my notebook? Yes!
* Result: Return index of 2 (which is 0) and current index (which is 1). Result: [0, 1].

Short answer: No, complement is a value (a number), not an index.
However, we use that value to look up an index in our dictionary. Here is the distinction:
## 1. The Value (complement)
complement is just the "missing piece" of math.

* If your Target is 10 and your current Number is 7...
* The complement is 3 ($10 - 7 = 3$).
* It is a pure number, not a position.

## 2. The Index (num_to_index[complement])
The dictionary (num_to_index) is like a map that links Values to Indices.

* Key: The number we saw before (e.g., 3).
* Value: Where we saw it (e.g., Index 0).

------------------------------
## See it in action:
Look at line 8 in the code:
return [num_to_index[complement], i]

* complement: "I am looking for the number 3."
* num_to_index[complement]: "The notebook says the number 3 was found at Index 0."
* i: "I am currently at Index 1."

So the code returns [0, 1].
