```python
class Node:

  def __init__(self,value):
    self.data=value
    self.next=None
```

```python
class Stack:

  def __init__(self):
    self.top=None

  def is_empty(self):
    return self.top==None

  def push(self,value):
    new_node=Node(value)
    new_node.next=self.top
    self.top=new_node

  def traverse(self):
    curr=self.top

    if curr==None:
      print("Stack is empty")
      return

    while curr!=None:
      print(curr.data)
      curr=curr.next

  def peek(self):
    if self.is_empty():
      return "stack is empty"
    return self.top.data

  def pop(self):
    if (self.is_empty()):
      return "stack is empty"
    else:
      data=self.top.data
      self.top=self.top.next
      return data

  def len(self):
    curr=self.top
    counter=0

    while curr!=None:
      counter+=1
      curr=curr.next
    return counter
```

```python
def reverse_string(text):

  s=Stack()
  for i in text:
    s.push(i)

  preserved=""
  while s.is_empty()==False:
    preserved+=s.pop()

  print(preserved)
```
