```python
class Node:

  def __init__(self,value):
    self.data=value
    self.next=None
```

```python
class Stack:

  def __init__(self):
    self.top=None

  def is_empty(self):
    return self.top==None

  def push(self,value):
    new_node=Node(value)
    new_node.next=self.top
    self.top=new_node

  def traverse(self):
    curr=self.top

    if curr==None:
      print("Stack is empty")
      return

    while curr!=None:
      print(curr.data)
      curr=curr.next

  def peek(self):
    if self.is_empty():
      return "stack is empty"
    return self.top.data

  def pop(self):
    if (self.is_empty()):
      return "stack is empty"
    else:
      data=self.top.data
      self.top=self.top.next
      return data

  def len(self):
    curr=self.top
    counter=0

    while curr!=None:
      counter+=1
      curr=curr.next
    return counter

  def size(self):
    return self.len()
```

```python
def reverse_string(text):

  s=Stack()
  for i in text:
    s.push(i)

  preserved=""
  while s.is_empty()==False:
    preserved+=s.pop()

  print(preserved)
```

```python
def text_editor(text,pattern):

  u=Stack()
  r=Stack()
  f=Stack()

  for i in text:
    u.push(i)

  for i in pattern:
    if i=='u':
      r.push(u.pop())
    elif i=='r':
      u.push(r.pop())

  while u.is_empty()==False:
    f.push(u.pop())

  while f.is_empty()==False:
    print(f.pop(),end="")
```

```python
def find_the_celeb(L):
  s=Stack()

  for i in range(len(L)):
    s.push(i)

  while s.size()>=2:

    i=s.pop()
    j=s.pop()

    if L[i][j]==0:
      s.push(i)
    else:
      s.push(j)

  celeb=s.pop()

  for i in range(len(L)):
    if i!=celeb:

      if L[i][celeb]==0 or L[celeb][i]==1:
        print("No celebrity")
        return
  print("The celebrity is",celeb)
```
