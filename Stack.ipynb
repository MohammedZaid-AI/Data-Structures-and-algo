```python
class Node:

  def __init__(self,value):
    self.data=value
    self.next=None
```

```python
class Stack:

  def __init__(self):
    self.top=None

  def is_empty(self):
    return self.top==None

  def push(self,value):
    new_node=Node(value)
    new_node.next=self.top
    self.top=new_node

  def traverse(self):
    curr=self.top

    if curr==None:
      print("Stack is empty")
      return

    while curr!=None:
      print(curr.data)
      curr=curr.next
```

```python
s=Stack()
```

```python
s.is_empty()
```

```python
s.push(2)
s.push(3)
s.push(4)
```
