```python
class Node:

  def __init__(self,value):
    self.data=value
    self.next=None
```

```python
class Queue:

  def __init__(self):
    self.front=None
    self.rear=None

  def enqueue(self,value):
    new_node=Node(value)

    if self.rear==None:
      self.rear=new_node
      self.front=self.rear
    else:
      self.rear.next=new_node
      self.rear=new_node

  def dequeue(self):
    if self.front==None:
      return "Queue is empty"
    else:
      self.front=self.front.next

  def traverse(self):
    if self.front==None:
      return "queue is empty"
    else:
      curr=self.front
      while curr!=None:
        print(curr.data,end=" ")
        curr=curr.next

  def is_empty(self):
    return self.front==None

  def size(self):
    curr=self.front
    counter=0

    while curr!=None:
      curr=curr.next
      counter+=1
    return counter


  def front_item(self):
    if self.front==None:
      return "Queue is empty"
    else:
      return self.front.data

  def tail_item(self):
    if self.rear==None:
      return "Queue is empty"
    else:
      return self.rear.data
```

```python
q=Queue()
```

```python
q.enqueue(2)
q.enqueue(3)
q.enqueue(4)
q.enqueue(5)
```

```python
q.traverse()
```

```python
q.size()
```

```python
q.front_item()
```
