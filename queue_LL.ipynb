```python
class Node:
  def __init__(self,value):
    self.data=value
    self.next=None
```

```python
class Queue:

  def __init__(self):
    self.front=None
    self.rear=None

  def enqueue(self,value):

    new_node=Node(value)

    if self.rear==None:
      self.front=new_node
      self.rear=self.front
    else:
      self.rear.next=new_node
      self.rear=new_node

  def dequeue(self):
    if self.front==None:
      return "queue is empty"
    else:
      self.front=self.front.next

  def size(self):
    temp=self.front
    counter=0

    while temp!=None:
      counter+=1
      temp=temp.next
    return counter

  def traverse(self):
    temp=self.front

    while temp !=None:
      print(temp.data)
      temp=temp.next
```

```python
q=Queue()

q.enqueue(1)
q.enqueue(2)
q.enqueue(3)
q.enqueue(4)
q.enqueue(5)
q.enqueue(6)
```

```python
def twostack_queue(item, action='enqueue'):

  if action == "enqueue":
    s1.push(item)
  else:
    if s2.is_empty():
      while not s1.is_empty():
        s2.push(s1.peek())
        s1.pop()
    s2.pop()


```
