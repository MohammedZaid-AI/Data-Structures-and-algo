```python
arr=[10, 20, 30, 40, 50]
target=20
mid=len(arr)//2
left=0
right=len(arr)-1
```

```python
def binary(arr,target,left,mid,right):
  if left> right:
    print("Not Found")
  elif arr[mid]==target:
    print("Found")
  elif target>arr[mid]:
    left=mid+1
    mid=(left + right)//2
    binary(arr,target,left,mid,right)
  elif target<arr[mid]:
    right=mid-1
    mid=(left+right)//2
    binary(arr,target,left,mid,right)
  else:
    print("Not Found")

```

```python
def fibbonacci(n):
  if n==0:
    return 0
  elif n==1:
    return 1
  return fibbonacci(n-1)+fibbonacci(n-2)
```
