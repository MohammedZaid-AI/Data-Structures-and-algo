```python
class Node:

  def __init__(self,data):
    self.data=data
    self.next=None

```

```python
class Stack:
  def __init__(self):
    self.top=None

  def is_empty(self):
    return self.top==None

  def push(self,value):
    new_node=Node(value)
    new_node.next=self.top
    self.top=new_node

  def traverse(self):

    temp=self.top
    while temp!=None:
      print(temp.data)
      temp=temp.next

  def peek(self):
    if self.top==None:
      return "Stack is empty"
    else:
      return self.top.data

  def pop(self):
    if self.top==None:
      return "stack is empty"
    else:
      data=self.top.data
      self.top=self.top.next
      return data

  def size(self):
    temp=self.top
    counter=0

    while temp!=None:
      temp=temp.next
      counter+=1
    return counter
```

```python
def reverse(value):
    s=Stack()
    for i in value:
      s.push(i)

    rev=""
    while not s.is_empty():
      rev= rev +s.pop()
    print(rev)
```

```python
def text_editor(text,pattern):

  u=Stack()
  r=Stack()

  for i in text:
    u.push(i)

  for i in pattern:
    if i=="u":
      data=u.pop()
      r.push(data)
    else:
      data=r.pop()
      u.push(data)

  res=''
  while not u.is_empty():
    res=u.pop()+res

  print(res)
```

```python
def check_celeb(l):
  s=Stack()

  for i in range(len(l)):
    s.push(i)

  while s.size()>=2:

    i=s.pop()
    j=s.pop()

    if l[i][j]==0:
      s.push(i)
    else:
      s.push(j)

  celeb=s.pop()

  for i in range(len(l)):
    if i!=celeb:
      if l[i][celeb]==0 or l[celeb][i]==1:
        print("no Celeb found")
        return

  print("the celeb is",celeb)




```

```python
def paranthesis_check(text):
  s=Stack()

  for i in text:
    if i=="(" or i=="[" or i=="{":
      s.push(i)

    elif i==")" or i=="]" or i=="}":
      if s.is_empty():
        return False
      elif i==")" and s.peek()=="(" or i=="]" and s.peek()=="[" or i=="}" and s.peek()=="{":
        s.pop()
      else:
        return False
  return s.is_empty()
```
